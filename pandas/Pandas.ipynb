```python
#Using **Pandas Series**, perform the following:

#✅ Create a Series named `scores` using the dictionary `student_scores`.

#✅ Print the marks scored by **"Charlie"** and **"Eva"**.

#✅ Add **5 bonus marks** to all students and print the updated Series.

#✅ Display only the students who scored **above 90** after the bonus.

#✅ Calculate and print the **average marks** of the class (after the bonus).

import pandas as pd
student_scores={'Alice':80,'Bob':75,'Charlie':90,'David':65,'Eva':98}
scores=pd.Series(student_scores)
print(scores)
print(scores['Charlie'])
print(scores['Eva'])
scores+=5
print(scores)
print(scores[scores>90])
print(scores.mean())

```

```python
from google.colab import drive
import pandas as pd

# Mount drive
drive.mount('/content/drive')

# Load dataset
df = pd.read_csv('/content/drive/My Drive/OnlineRetail.csv', encoding='ISO-8859-1')

# Strip spaces from column names (good practice)
df.columns = df.columns.str.strip()
print(df.columns.tolist())  # verify columns

# Inspect
print(df.head())
print(df.tail())
df.info()
print(df.shape)
print(df.dtypes)

# Missing data handling
df = df.dropna(subset=['CustomerID'])          # Drop rows with missing CustomerID
df['Description'] = df['Description'].fillna("Unknown")  # Fill missing descriptions

# Convert InvoiceDate to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')
df = df.dropna(subset=['InvoiceDate'])  # drop any rows where date conversion failed

#treat bad data
returns=df[df['Quantity']<0]
returns.head()

df=df[df['UnitPrice']>0]

#Duplicates
df.duplicated()
df=df.drop_duplicates()

#Filtering & Sorting
sus=df[df['Quantity']>1000]
df['revenue']=df['UnitPrice']*df['Quantity']
df=df.sort_values(by='revenue',ascending=False)

#Merge/Enrich data
product_info = pd.DataFrame({
    'StockCode': ['85123A', '71053', '84406B'],
    'Category': ['Home Decor', 'Home Decor', 'Home Decor'],
    'Supplier': ['Supplier1', 'Supplier2', 'Supplier1']
})
df=df.merge(product_info,on='StockCode',how='left')


#Aggregation
country_agg = df.groupby('Country').agg(
    total_revenue=pd.NamedAgg(column='Revenue', aggfunc='sum'),
    mean_revenue=pd.NamedAgg(column='Revenue', aggfunc='mean'),
    num_orders=pd.NamedAgg(column='InvoiceNo', aggfunc='nunique')
)


# Bucket UnitPrice using pd.cut()
price_bins = [0, 5, 20, df['UnitPrice'].max()]
price_labels = ['cheap', 'medium', 'expensive']
df['PriceRange'] = pd.cut(df['UnitPrice'], bins=price_bins, labels=price_labels)

# Group by PriceRange
price_agg = df.groupby('PriceRange').agg(
    num_purchases=pd.NamedAgg(column='InvoiceNo', aggfunc='count'),
    avg_revenue=pd.NamedAgg(column='Revenue', aggfunc='mean')
).reset_index()
print("\nAggregated by price range:\n", price_agg)

# =========================
# 4️⃣ Correlation / Relationships
# =========================
numeric_cols = ['Quantity', 'UnitPrice', 'Revenue']
corr_matrix = df[numeric_cols].corr()
print("\nCorrelation matrix:\n", corr_matrix)

# =========================
# 5️⃣ Rename & String Operations
# =========================
df = df.rename(columns={'StockCode': 'Product_ID', 'UnitPrice': 'Price'})

# Lowercase descriptions
df['Description_lower'] = df['Description'].str.lower()

# Filter products containing 'glass'
glass_products = df[df['Description_lower'].str.contains('glass')]
print("\nProducts containing 'glass':\n", glass_products.head())
```
